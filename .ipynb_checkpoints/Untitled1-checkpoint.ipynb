```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt

# 1. GENERATE SAMPLE DATA (14 Days of Hourly Visitor Counts)
np.random.seed(42)
dates = pd.date_range(start="2026-08-01", periods=336, freq="h")
hours = dates.hour

# Base traffic pattern: daytime peaks, night troughs + noise
visitors = 50 + 40 * np.sin((hours - 6) * np.pi / 12) + np.random.normal(0, 5, 336)
visitors = np.asarray(np.clip(visitors, 0, None))

# Inject two obvious anomalies
visitors[100] = 200  # Extreme high spike
visitors[220] = 0    # Sudden drop during peak hours

df = pd.DataFrame({"timestamp": dates, "visitors": visitors})

# 2. FEATURE ENGINEERING
# Time features help the tree understand time-of-day context
df["hour"] = df["timestamp"].dt.hour
df["dayofweek"] = df["timestamp"].dt.dayofweek

features = ["visitors", "hour", "dayofweek"]
X = df[features]

# 3. FIT ISOLATION FOREST
# contamination=0.01 expects ~1% of the data to be anomalous
model = IsolationForest(
    n_estimators=100,
    contamination=0.01,
    random_state=42
)

# Fit and predict: 1 for normal (inliers), -1 for anomalies (outliers)
df["anomaly_label"] = model.fit_predict(X)
df["anomaly_score"] = model.decision_function(X) # Lower scores = more anomalous

# Filter detected anomalies
anomalies = df[df["anomaly_label"] == -1]

print("--- DETECTED ANOMALIES ---")
print(anomalies[["timestamp", "visitors", "anomaly_score"]])

# 4. VISUALIZATION
plt.figure(figsize=(12, 4))
plt.plot(df["timestamp"], df["visitors"], label="Visitor Count", alpha=0.7)
plt.scatter(
    anomalies["timestamp"], 
    anomalies["visitors"], 
    color="red", 
    label="Anomaly (-1)", 
    zorder=5
)
plt.title("Isolation Forest Anomaly Detection (sklearn)")
plt.xlabel("Timestamp")
plt.ylabel("Visitors")
plt.legend()
plt.tight_layout()
plt.show()
```
