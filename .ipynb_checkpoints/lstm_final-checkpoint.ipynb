```python
import os
import random
import numpy as np
import pandas as pd
from datetime import timedelta
from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential,Model
from tensorflow.keras.layers import LSTM,Dense,Input
from tensorflow.keras.callbacks import EarlyStopping
```

```python
#Load the cleaned data

data = pd.read_csv('final_data.csv')
data.dtypes
print(len(data))
```

```python
#Change timestamp into datetime

data['timestamp'] = pd.to_datetime(data['timestamp'],utc=False)

data.dtypes

#Time is now in Central European Summer Time
```

```python
#Configuration for determinstic result

seed = 42
os.environ['PYTHONHASHSEED'] = str(seed)
random.seed(seed)
np.random.seed(seed)
tf.random.set_seed(seed)
tf.config.experimental.enable_op_determinism()
```

```python
#Configuration for training
extraction_features = ['SID','timestamp','visitors','hour_cos','hour_sin','weekday_sin','weekday_cos']
features = ['visitors','hour_cos','hour_sin','weekday_sin','weekday_cos']
n_features = len(features)
window_size = 24
```

```python
#Seperate 12 days of data of each sensor for training and 2 days for testing

start_date = data['timestamp'].min()
end_date = data['timestamp'].max()
two_day_before = end_date - timedelta(days=2)
test_timestamp = pd.date_range(start= two_day_before, end=end_date,freq = 'h')
train_timestamp = pd.date_range(start= start_date, end= two_day_before-timedelta(hours=1),freq = 'h')

test_data = data[data['timestamp'].isin(test_timestamp)][extraction_features]
train_data = data[data['timestamp'].isin(train_timestamp)][extraction_features]

#Check either the data splitting is true
assert(len(data) == len(test_data) + len(train_data))
```

```python
def create_windows(data,window_size):
    x_vals= []
    for i in range(len(data) - window_size):
        x_vals.append(data[i:i+window_size])
    return np.array(x_vals)

def extract_target(data,window_size,target_column):
    y_vals= []
    for i in range(len(data) - window_size):
        y_vals.append(data[i+window_size,target_column])
    
    return np.array(y_vals)
    
```

```python
#Standardizing Data
# Data must be standardized according to the sensors

sensors = list(data['SID'].unique())

X_train = []
y_train = []
X_test = []
y_test = []

for sensor in sensors: 

    sensor_train_data = train_data[train_data['SID'] == sensor][features].values
    sensor_test_data = test_data[test_data['SID'] == sensor][features].values
    y_train_sensor = extract_target(sensor_train_data,24,0)
    y_test_sensor = extract_target(sensor_test_data,24,0)
    scaler = StandardScaler()
    sensor_train_data_scaled = scaler.fit_transform(sensor_train_data)
    sensor_test_data_scaled = scaler.transform(sensor_test_data)

    X_train_sensor = create_windows(sensor_train_data_scaled,24)
    X_test_sensor= create_windows(sensor_test_data_scaled,24)

    X_train.append(X_train_sensor)
    y_train.append(y_train_sensor)
    X_test.append(X_test_sensor)
    y_test.append(y_test_sensor)


X_train = np.concatenate(X_train, axis=0)
y_train = np.concatenate(y_train, axis=0).reshape(-1,1)

X_test = np.concatenate(X_test, axis=0)
y_test = np.concatenate(y_test, axis=0).reshape(-1,1)



    

       
    

```

```python
def loss_function_analysis(history):
    history_dict = history.history
    train_loss = history_dict['loss']
    #validation_loss = history_dict['val_loss']
    epoches = [i for i in range(1,len(train_loss)+1)]
    plt.plot(epoches,train_loss,label='Train Loss',color = 'black')
    #plt.plot(epoches,validation_loss,label = 'Validation Loss', color = 'red')
    plt.xlabel('Epoches')
    plt.ylabel('Loss')
    plt.title('Loss Function Analysis')
    plt.legend()
    plt.show()
```

```python
def model():
    inputs = Input(shape=(24,5))
    x = LSTM(64, return_sequences=True, name='Lstm_layer1')(inputs)
    x = LSTM(32,return_sequences=True, name = 'Lstm_layer2')(x)
    x = LSTM(16, return_sequences=False,name='Lstm_layer3')(x)
    x = Dense(8,activation='relu',name='Dense_layer')(x)
    outputs = Dense(1,name = 'Output_layer')(x)

    return keras.models.Model(inputs,outputs)
```

```python
lstm_model = model()

lstm_model.compile(
    optimizer = 'adam',
    loss = 'mse'
)

lstm_model.summary()
```

```python
early_stop = EarlyStopping(
    monitor = 'loss',
    patience = 5,
    min_delta= 1e-2,
    restore_best_weights= True
)

hisotry = lstm_model.fit(
    X_train,y_train,
    epochs = 81,
    batch_size = 64
)
```

```python
loss_function_analysis(hisotry)
```

```python
y_preds = lstm_model.predict(X_test)
```
