# Global Anomaly Detection Model (v2)

Changes from the original `globalModel_final.ipynb`:
- Timestamps are parsed to `datetime` **before** sorting (avoids incorrect ordering if the raw strings aren't lexically sortable).
- A chronological **train/validation split per sensor** is used, with `StandardScaler` fit only on the training portion (no leakage from the held-out period).
- `EarlyStopping` monitors `val_loss` instead of `loss`, so it actually reflects generalization.
- `NUM_SENSORS` is derived from the data instead of hardcoded.
- Anomaly scoring runs over the full timeline (train + validation) using the model trained only on the earlier portion.
- Removed dead/commented-out code, fixed `SettingWithCopyWarning` risks, fixed the mismatched Kurpark/Innenstadt cell at the end, and cleaned up naming typos.

```python
#Load libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import os
import random
import tensorflow as tf
import seaborn as sns
from tensorflow import keras
from tensorflow.keras.layers import (Input, LSTM, RepeatVector, TimeDistributed, Dense, Embedding, Flatten,
                                      Concatenate, Dropout)
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.preprocessing import StandardScaler, LabelEncoder
from collections import defaultdict
```

```python
#Load the cleaned data
data = pd.read_csv('final_data.csv')
print(f"Loaded {len(data)} rows")
data.columns
```

## Configuration

```python
SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
tf.config.experimental.enable_op_determinism()

continuous_features = ['visitors']
cyclic_features = ['hour_sin', 'hour_cos', 'weekday_sin', 'weekday_cos']
features = continuous_features + cyclic_features
n_features = len(features)
window_size = 24

# Fraction of each sensor's timeline (in chronological order) used for training.
# The remaining tail is held out purely for early stopping (val_loss), not for
# tuning any other hyperparameters.
TRAIN_FRACTION = 0.85
```

**Fix:** parse `timestamp` to `datetime` *before* sorting. Sorting on the raw string
column can silently produce the wrong chronological order depending on the source format.

```python
data['timestamp'] = pd.to_datetime(data['timestamp'])
data = data.sort_values(['SID', 'timestamp']).reset_index(drop=True)
```

## Build sliding windows per sensor

For each sensor:
- Split its rows chronologically into a training portion (`TRAIN_FRACTION`) and a held-out validation tail.
- Fit the `StandardScaler` **only on the training portion**, then apply it to the full series. This avoids leaking
  statistics from the validation period into the scaling used for the whole timeline.
- Build sliding windows over the *entire* series (`all_*`) for later anomaly scoring, while separately tracking which
  windows fall entirely inside the training portion (`train_*`) vs. the validation tail (`val_*`) for model fitting.

```python
scalers = {}

all_windows, all_sensor_ids, all_window_times, all_location = [], [], [], []
train_windows, train_sensor_ids = [], []
val_windows, val_sensor_ids = [], []

for sensor in data['SID'].unique():
    sensor_data = data[data['SID'] == sensor].reset_index(drop=True)
    n_rows = len(sensor_data)

    train_cutoff = int(n_rows * TRAIN_FRACTION)

    # Fit the scaler ONLY on the training portion.
    scaler = StandardScaler()
    scaler.fit(sensor_data.loc[:train_cutoff - 1, continuous_features])
    scalers[sensor] = scaler

    sensor_data = sensor_data.copy()
    sensor_data[continuous_features] = scaler.transform(sensor_data[continuous_features])

    values = sensor_data[features].values
    timestamps = sensor_data['timestamp'].values
    names = sensor_data['name'].values

    n_windows = n_rows - window_size + 1
    # A window counts as "train" only if every row it spans is inside the train portion.
    train_window_count = max(train_cutoff - window_size + 1, 0)

    for i in range(n_windows):
        window = values[i:i + window_size]
        end_idx = i + window_size - 1

        all_windows.append(window)
        all_sensor_ids.append(sensor)
        all_window_times.append(timestamps[end_idx])
        all_location.append(names[end_idx])

        if i < train_window_count:
            train_windows.append(window)
            train_sensor_ids.append(sensor)
        else:
            val_windows.append(window)
            val_sensor_ids.append(sensor)

all_windows = np.array(all_windows)
train_windows = np.array(train_windows)
val_windows = np.array(val_windows)

all_sensor_ids = np.array(all_sensor_ids)
train_sensor_ids = np.array(train_sensor_ids)
val_sensor_ids = np.array(val_sensor_ids)

print(f"All windows:   {all_windows.shape}")
print(f"Train windows: {train_windows.shape}")
print(f"Val windows:   {val_windows.shape}")
```

**Fix:** fit the `LabelEncoder` on *all* sensor ids up front, and derive `NUM_SENSORS` from it,
instead of hardcoding the sensor count.

```python
encoder = LabelEncoder()
encoder.fit(all_sensor_ids)

all_sensor_id_encoded = encoder.transform(all_sensor_ids).reshape(-1, 1)
train_sensor_id_encoded = encoder.transform(train_sensor_ids).reshape(-1, 1)
val_sensor_id_encoded = encoder.transform(val_sensor_ids).reshape(-1, 1)

NUM_SENSORS = len(encoder.classes_)
print(f"NUM_SENSORS = {NUM_SENSORS}")
```

```python
#Plot the loss functions
def loss_function_analysis(history):
    history_dict = history.history
    train_loss = history_dict['loss']
    val_loss = history_dict.get('val_loss')
    epochs = range(1, len(train_loss) + 1)

    plt.plot(epochs, train_loss, label='Train Loss', color='black')
    if val_loss is not None:
        plt.plot(epochs, val_loss, label='Validation Loss', color='red')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Loss Function Analysis')
    plt.legend()
    plt.show()
```

## Model definition (unchanged architecture, dynamic `NUM_SENSORS`)

```python
EMBED_DIM = 6

inputs = Input(shape=(window_size, n_features), name='Input_Layer')
sensor_input = Input(shape=(1,), dtype='int32', name='Sensor_ID_Input')

# --- Sensor embedding, repeated across all timesteps ---
sensor_embed = Embedding(NUM_SENSORS, EMBED_DIM, name='sensor_embedding')(sensor_input)
sensor_embed = Flatten()(sensor_embed)                              # (batch, EMBED_DIM)
sensor_embed_seq = RepeatVector(window_size)(sensor_embed)          # (batch, window_size, EMBED_DIM)

# --- Concatenate sensor embedding to the raw input at every timestep ---
x = Concatenate(axis=-1, name='concat_sensor_input')([inputs, sensor_embed_seq])

# --- Encoder ---
x = LSTM(32, return_sequences=True, name='encoder1')(x)
x = LSTM(64, return_sequences=False, name='encoder2')(x)

# --- Bottleneck, also conditioned on sensor ---
x = Concatenate(name='concat_sensor_latent')([x, sensor_embed])
latent = Dense(16, activation='relu', name='latent')(x)
latent_dropped = Dropout(0.2)(latent)

# --- Decoder ---
x = RepeatVector(window_size)(latent_dropped)
x = LSTM(32, return_sequences=True, name='decoder1')(x)
x = LSTM(64, return_sequences=True, name='decoder2')(x)
outputs = TimeDistributed(Dense(n_features))(x)

model = keras.models.Model(inputs=[inputs, sensor_input], outputs=outputs)
model.summary()
```

```python
model.compile(optimizer='adam', loss='mse')
```

**Fix:** train with an explicit `validation_data` set and monitor `val_loss` for early stopping,
so stopping actually reflects generalization rather than training-set fit.

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    min_delta=1e-2,
    restore_best_weights=True
)

history = model.fit(
    [train_windows, train_sensor_id_encoded], train_windows,
    validation_data=([val_windows, val_sensor_id_encoded], val_windows),
    epochs=20,
    batch_size=64,
    callbacks=[early_stop],
)
```

```python
loss_function_analysis(history)
```

## Score every window for anomaly detection

Inference runs over **all** windows (train + validation), using the model trained only on the earlier portion of each
sensor's history. This lets the validation tail act as a genuine out-of-sample check on reconstruction error, not just
on training loss.

```python
pred = model.predict([all_windows, all_sensor_id_encoded])
```

```python
# Scaled squared error, per timestep -- a quick training-convergence sanity check only.
# NOT used for the anomaly thresholds below (those use unscaled absolute error, next).
scaled_point_errors = np.square(all_windows[:, :, 0] - pred[:, :, 0])
print(f"Scaled error (diagnostic) -- mean: {scaled_point_errors.mean():.4f}, "
      f"std: {scaled_point_errors.std():.4f}, max: {scaled_point_errors.max():.4f}")
```

Inverse-transform predictions and actuals back to real visitor counts (using each sensor's own scaler) and compute
point-wise absolute error, averaged across all overlapping windows for each (sensor, timestamp).

```python
point_error_map = defaultdict(list)
timestamp_lookup = {}

window_idx = 0
for sensor in data['SID'].unique():
    sensor_data = data[data['SID'] == sensor].reset_index(drop=True)
    timestamps = sensor_data['timestamp'].values
    scaler = scalers[sensor]
    n_windows = len(sensor_data) - window_size + 1

    for i in range(n_windows):
        pred_scaled = pred[window_idx, :, 0].reshape(-1, 1)
        actual_scaled = all_windows[window_idx, :, 0].reshape(-1, 1)

        pred_unscaled = scaler.inverse_transform(pred_scaled).flatten()
        actual_unscaled = scaler.inverse_transform(actual_scaled).flatten()

        unscaled_errors = np.abs(actual_unscaled - pred_unscaled)

        for j in range(window_size):
            position = i + j
            point_error_map[(sensor, position)].append(unscaled_errors[j])
            timestamp_lookup[(sensor, position)] = timestamps[position]

        window_idx += 1

point_results = []
for (sensor, position), errors in point_error_map.items():
    location_name = data.loc[data['SID'] == sensor, 'name'].iloc[0]
    point_results.append({
        'SID': sensor,
        'location': location_name,
        'timestamp': timestamp_lookup[(sensor, position)],
        'point_error': np.mean(errors),
    })

point_results = pd.DataFrame(point_results)
point_results.head()
```

```python
point_results.sort_values('point_error', ascending=False).head(20)[
    ['SID', 'timestamp', 'point_error', 'location']
]
```

```python
def plot_error_distribution(sensor_id):
    sensor_errors = point_results.loc[point_results['SID'] == sensor_id, 'point_error']
    location = point_results.loc[point_results['SID'] == sensor_id, 'location'].iloc[0]
    sns.histplot(sensor_errors, bins=50)
    plt.title(f"Error distribution at {location}")
    plt.show()
```

```python
sensors = list(point_results['SID'].unique())

for sensor in sensors:
    plot_error_distribution(sensor)
```

Per-sensor threshold via IQR (`Q3 + 1.5 * IQR`) on the mean point error. This is a standard, easy-to-explain
outlier rule; consider a percentile-based or rolling-window threshold if you find IQR too sensitive/insensitive
for a given sensor's error distribution.

```python
point_thresholds = {}
for sensor in point_results['SID'].unique():
    errors = point_results.loc[point_results['SID'] == sensor, 'point_error']
    Q1 = np.percentile(errors, 25)
    Q3 = np.percentile(errors, 75)
    IQR = Q3 - Q1
    point_thresholds[sensor] = Q3 + 1.5 * IQR
```

```python
point_results.to_csv('point_results.csv', index=False)
```

**Fix:** use `.copy()` on the filtered slice before mutating it, to avoid `SettingWithCopyWarning`
(and the associated risk of silently not writing back to the DataFrame).

```python
def plot_anomalies(sensor_id):
    sensor_data = point_results[point_results['SID'] == sensor_id].copy()
    sensor_data['timestamp'] = pd.to_datetime(sensor_data['timestamp'])
    sensor_data = sensor_data.sort_values('timestamp')
    location = sensor_data['location'].iloc[0]
    threshold = point_thresholds[sensor_id]

    anomalies = sensor_data[sensor_data['point_error'] > threshold]

    plt.figure(figsize=(12, 7))
    plt.plot(sensor_data['timestamp'], sensor_data['point_error'], label='error', color='black', linewidth=1)
    plt.axhline(threshold, color='red', linestyle='--', label=f'Threshold: {threshold:.2f}')
    plt.scatter(anomalies['timestamp'], anomalies['point_error'], color='red', s=10, marker='o',
                edgecolor='red', facecolor='white', label=f'anomaly -> count {len(anomalies)}')
    plt.xlabel('Date')
    plt.ylabel('Reconstruction Error')
    plt.title(f'Anomalies at {location}')
    plt.xticks(rotation=45)
    plt.legend(loc='upper right')
    plt.tight_layout()
    plt.show()
```

```python
for sensor in sensors:
    plot_anomalies(sensor)
```

```python
def prepare_anomaly_report(sensor_id):
    directory = Path('anomaly_report')
    directory.mkdir(parents=True, exist_ok=True)

    sensor_data = point_results[point_results['SID'] == sensor_id].copy()
    location = sensor_data['location'].iloc[0]
    threshold = point_thresholds[sensor_id]
    sensor_data['anomaly'] = np.where(sensor_data['point_error'] > threshold, 'Yes', 'No')

    output_file = directory / f"{location}.csv"
    sensor_data.to_csv(output_file, index=False)
```

```python
for sensor in sensors:
    prepare_anomaly_report(sensor)
```

```python
#Save model for further use
model.save('global_model.keras')
```
