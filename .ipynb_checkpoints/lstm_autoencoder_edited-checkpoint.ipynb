```python
#Load libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import os
import random
import tensorflow as tf
import seaborn as sns
from tensorflow import keras
from tensorflow.keras.layers import (Input, LSTM, RepeatVector,TimeDistributed, Dense,Embedding, Flatten,
Concatenate)
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.preprocessing import LabelEncoder

```

```python
#Load the cleaned data
data = pd.read_csv('final_data.csv')
print(len(data))
data.columns
```

```python
#Configurations

SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
tf.config.experimental.enable_op_determinism()

features = ['visitors','hour_cos','hour_sin','weekday_sin','weekday_cos']
extracted_features = ['SID','name','timestamp','visitors','hour_cos','hour_sin','weekday_sin','weekday_cos']
n_features = len(features)
window_size = 24
```

```python
data = data.sort_values(['SID','timestamp'])
data['timestamp'] = pd.to_datetime(data['timestamp'])

print(data)
```

```python
def create_window(data):
    X = []
    y = []
    window_times = []
    location = []

    for sensor in data['SID'].unique():
        sensor_data = data[data['SID'] == sensor]

        values = sensor_data[features].to_numpy(dtype=np.float32)
        timestamps = sensor_data['timestamp'].values
        name = sensor_data['name'].values

        for i in range(len(values) - window_size):
            X.append(values[i:i+window_size])
            window_times.append(timestamps[i+window_size-1])
            location.append(name[i+window_size-1])

    return (
        np.array(X),
        np.array(window_times),
        location,
    )

train_data,train_times,train_location = create_window(train_data) 
validation_data,val_times,val_location = create_window(val_data) 
print(train_data.shape) 
print(validation_data.shape) 
```

```python
#Standardarizing the data
scaler = StandardScaler()
train_data_reshaped = train_data.reshape(-1, n_features)
train_data_scaled = scaler.fit_transform(train_data_reshaped)
train_data = train_data_scaled.reshape(-1, window_size, n_features)

validation_data_reshaped = validation_data.reshape(-1, n_features)
validation_data_scaled = scaler.transform(validation_data_reshaped)
validation_data = validation_data_scaled.reshape(-1, window_size, n_features)

print(train_data.shape)
print(validation_data.shape)
```

```python
all_data = []

for sensor in data['SID'].unique():

    sensor_data = data[data['SID'] == sensor]

    values = sensor_data[features].values
    for i in range(len(values) - window_size + 1):
        all_data.append(values[i:i+window_size])

all_data = np.array(all_data)


all_data_reshaped = all_data.reshape(-1, n_features)
all_data_scaled = scaler.transform(all_data_reshaped)
all_data = all_data_scaled.reshape(-1, window_size, n_features)
print(all_data.shape)
```

```python
#Plot the loss functions

def loss_function_analysis(history):
    history_dict = history.history
    train_loss = history_dict['loss']
    validation_loss = history_dict['val_loss']
    epoches = [i for i in range(1,len(train_loss)+1)]
    plt.plot(epoches,train_loss,label='Train Loss',color = 'black')
    plt.plot(epoches,validation_loss,label = 'Validation Loss', color = 'red')
    plt.xlabel('Epoches')
    plt.ylabel('Loss')
    plt.title('Loss Function Analysis')
    plt.legend()
    plt.show()
```

```python

inputs = Input(shape=(window_size, n_features), name='Input_Layer')

#Encoder
x = LSTM(64, return_sequences=True, name='encoder1')(inputs)
x = LSTM(32,return_sequences=False,name='encoder2')(x)

# 16 dimensional Latent Representation
latent = Dense(16, activation='relu', name='latent')(x)

#Repeat vector
x = RepeatVector(window_size)(latent)

#Decoder
x = LSTM(32, return_sequences=True, name='decoder1')(x)
x = LSTM(64,return_sequences=True,name='decoder2')(x)
outputs = TimeDistributed(Dense(n_features))(x)

model =keras.models.Model(inputs=inputs, outputs=outputs)
```

```python
model.summary()
```

```python
model.compile(optimizer = 'adam',loss= 'mse')
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

history = model.fit(train_data,train_data,
          epochs =50,
          batch_size=32,
          validation_data= (validation_data,validation_data),
        callbacks= [early_stop])
```

```python
loss_function_analysis(history)
```

```python
pred = model.predict(all_data)
```

```python
# pred shape: (n_windows, window_size, n_features)
# get squared error at every timestep in every window:
point_errors = np.square(all_data[:, :, 0] - pred[:, :, 0])  # shape: (n_windows, window_size)
print(point_errors.shape)

```

```python
from collections import defaultdict

# Store errors for each data point
point_error_map = defaultdict(list)
timestamp_lookup = {}

window = 0

# Go through each sensor separately
for sensor in data['SID'].unique():

    sensor_data = data[data['SID'] == sensor].reset_index(drop=True)
    timestamps = sensor_data['timestamp'].values
    location = sensor_data['name'].iloc[0]

    # Create windows
    for i in range(len(sensor_data) - window_size + 1):

        # Store the reconstruction error of each point
        for j in range(window_size):
            position = i + j

            point_error_map[(sensor, position)].append(point_errors[window, j])
            timestamp_lookup[(sensor, position)] = timestamps[position]
        window += 1

# Calculate average error for each point
point_results = []

for key in point_error_map:

    sensor = key[0]
    position = key[1]

    sensor_data = data[data['SID'] == sensor]
    location = sensor_data['name'].iloc[0]

    point_results.append({
        'SID': sensor,
        'location': location,
        'timestamp': timestamp_lookup[key],
        'point_error': np.mean(point_error_map[key])
    })

point_results = pd.DataFrame(point_results)
```

```python
def error_distribution(sensor_id):
    sensor_data = point_results[point_results['SID'] == sensor_id]
    sensor_errors = sensor_data['point_error']
    sns.histplot(sensor_errors,bins=50)
    plt.title(f"Error distribution at {sensor_data['location'].iloc[0]}")
    plt.show()
```

```python
sensors = list(point_results['SID'].unique())

#for sensor in sensors:
#    error_distribution(sensor)

error_distribution(sensors[1])
```

```python
point_thresholds = {}
for sensor in point_results['SID'].unique():
    errors = point_results[point_results['SID'] == sensor]['point_error']
    Q1 = np.percentile(errors,25)
    Q3 = np.percentile(errors,75)
    IQR = Q3 - Q1
    threshold = Q3 + 3* IQR
    point_thresholds[sensor] = threshold
```

```python
#Determine the reconstruction error distribution for each sensor
def anomaly_plot(sensor_id):
    sensor_data = point_results[point_results['SID'] == sensor_id]
    sensor_data['timestamp'] = pd.to_datetime(sensor_data['timestamp'])
    sensor_data = sensor_data.sort_values('timestamp')
    location = sensor_data['location'].iloc[0]
    threshold = point_thresholds[sensor_id]
    plt.clf()
    plt.close()
    plt.figure(figsize=(12, 7))
    plt.plot(sensor_data['timestamp'], sensor_data['point_error'], label='error',color='black',linewidth=1)
    plt.axhline(threshold, color='red', linestyle='--', label= f'Threshold: {threshold: .2f}')
    anomalies = sensor_data[sensor_data['point_error'] > threshold]
    plt.scatter(anomalies['timestamp'], anomalies['point_error'], color='red', s=10,marker='o',edgecolor='red',
                facecolor = 'white',label= f'anomaly -> count {len(anomalies)} ')
    plt.xlabel('Date')
    plt.ylabel('Reconstruction Errors')
    plt.title(f'Anomaly on Sensor at {location}')
    plt.xticks(rotation=45)
    plt.legend(loc= 'upper right')
    plt.tight_layout()
    plt.show()
```

```python
#for sensor in sensors:
#    anomaly_plot(sensor)


anomaly_plot(sensors[1])
```

```python
def prepare_anamoly_report(sensor_id):
    directory = Path('anomaly_report')
    directory.mkdir(parents = True, exist_ok=True)
    
    sensor_data = point_results[point_results['SID'] == sensor_id]
    location = sensor_data['location'].iloc[0]
    threshold = point_thresholds[sensor_id]
    sensor_data['anomaly'] = np.where(
        sensor_data["point_error"] > threshold,
        "Yes",
        "No")

    output_file = directory/f"{location}.csv"
    sensor_data.to_csv(output_file)
```

```python
for sensor in sensors:
    prepare_anamoly_report(sensor)
```
