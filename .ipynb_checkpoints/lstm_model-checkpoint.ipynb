```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from keras import Layer
import os
import random
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense,Input
```

```python
#Load the cleaned data
data = pd.read_csv('final_data.csv')
print(len(data))
data.columns
```

```python
#Configurations
SEED = 42
os.environ['PYTHONHASHSEED'] = str(SEED)
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
tf.config.experimental.enable_op_determinism()

features = ['visitors','hour_cos','hour_sin','weekday_sin','weekday_cos']
n_features = len(features)
window_size = 24
```

```python
data = data.sort_values(['SID','timestamp'])
data['timestamp'] = pd.to_datetime(data['timestamp'])
```

```python
X = []
y = []
window_times = []
location = []
sensor_ids = []

for sensor in data['SID'].unique():
    sensor_data = data[data['SID'] == sensor]
    values = sensor_data[features].values
    timestamps = sensor_data['timestamp'].values
    name = sensor_data['name'].values

    for i in range(len(values)-window_size):
        X.append(values[i:i+window_size])
        y.append(values[i+window_size])
        window_times.append(timestamps[i + window_size - 1]) #Append the last time
        location.append(name[i + window_size - 1])


X = np.array(X)
y = np.array(y)
y = y[:,0]
y = y.reshape(-1,1)

print(X.shape)
```

```python
#Seperate Date for train and test
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=SEED,test_size=0.2)
```

```python
#Standardizing Data

scaler = StandardScaler()

train_data_reshaped = X_train.reshape(-1,n_features)
train_data_scaled = scaler.fit_transform(train_data_reshaped)
train_data = train_data_scaled.reshape(-1,window_size,n_features)

test_data_reshaped = X_test.reshape(-1,n_features)
test_data_scaled = scaler.fit_transform(test_data_reshaped)
test_data = test_data_scaled.reshape(-1,window_size,n_features)
```

```python
#Plot the loss functions
def loss_function_analysis(history):
    history_dict = history.history
    train_loss = history_dict['loss']
    validation_loss = history_dict['val_loss']
    epoches = [i for i in range(1,len(train_loss)+1)]
    plt.plot(epoches,train_loss,label='Train Loss',color = 'black')
    plt.plot(epoches,validation_loss,label = 'Validation Loss', color = 'red')
    plt.xlabel('Epoches')
    plt.ylabel('Loss')
    plt.title('Loss Function Analysis')
    plt.legend()
    plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

model = Sequential([
    
    # Input + first LSTM hidden layer
    LSTM(64, input_shape=(24, 5), return_sequences=True),

    # Second LSTM hidden layer
    LSTM(32, return_sequences=True),

    # Third LSTM hidden layer
    LSTM(16),

    # Dense hidden layer
    Dense(8, activation='relu'),

    # Output layer
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.summary()
```

```python
history = model.fit(X_train,y_train,epochs=48,
                   validation_data= (X_test,y_test))
```

```python
import matplotlib.pyplot as plt

loss_function_analysis(history)
```

```python
pred = model.predict(X_test)
```

```python
pred2 = model.predict(X_train)
```

```python
train_errors = np.abs(y_train - pred2 )

train_errors = np.log(train_errors)
```
